## Arquivo: cadastro_alunos.json

```python
df_Alunos_JSON = pd.read_json('https://github.com/FranciscoFoz/7_Days_of_Code_Alura-Python-Pandas/raw/main/Dia_6-Novos_dados_novas_analises/Datasets/cadastro_alunos.json')

# df_alunos_graduacao = pd.read_json(df_Alunos_JSON.registros[0])


df_Alunos_JSON
```

```python
import pandas as pd
import json

# Passo 1: Verificar a estrutura
print("Estrutura do DataFrame:")
print(df_Alunos_JSON.info())

# Passo 2: Acessar o registro corretamente 
# Filtrar somente os alunos de GRADUAÇÃO
registro = df_Alunos_JSON['registros'].iloc[0]  # ou .registros[0]

# Passo 3: Verificar o tipo
print(f"\nTipo do registro: {type(registro)}")

# Passo 4: Converter para DataFrame (escolha a opção adequada)
if isinstance(registro, str):
    # Se for string JSON
    dados = json.loads(registro)
    df_alunos_graduacao = pd.json_normalize(dados)
elif isinstance(registro, dict):
    # Se for dicionário
    df_alunos_graduacao = pd.json_normalize(registro)
elif isinstance(registro, list):
    # Se for lista
    df_alunos_graduacao = pd.DataFrame(registro)
else:
    print(f"Tipo inesperado: {type(registro)}")
```

```python
# Passo 5: Verificar resultado
print("\nDataFrame resultante:")
df_alunos_graduacao
```

```python
df_alunos_graduacao.dtypes
```
